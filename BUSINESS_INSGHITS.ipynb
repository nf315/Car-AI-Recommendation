1- Which Target has the highest average price?

```python
top = (
    df.groupby("Target")["Price"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12,6))
sns.barplot(x=top.values, y=top.index)

plt.title("Top 10 Targets by Average Price")
plt.xlabel("Average Price")
plt.ylabel("Target")
plt.show()
```

 2-Which category has the highest average price?

```python
cat = (
    df.groupby("Category")["Price"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(9,5))
sns.barplot(x=cat.values,y=cat.index)

plt.title("Average Price by Category")
plt.show()
```

3- Does mileage affect price?

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="Mileage",
    y="Price",
    alpha=.5
)

plt.title("Mileage vs Price")
plt.show()
```

4- Do newer cars cost more?

```python
year=df.groupby("Prod. year")["Price"].mean().reset_index()

plt.figure(figsize=(12,5))

sns.lineplot(
    data=year,
    x="Prod. year",
    y="Price",
    marker="o"
)

plt.title("Average Price by Production Year")
plt.show()
```

19-Does the number of airbags affect price?
